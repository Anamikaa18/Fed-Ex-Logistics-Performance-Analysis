## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Data Wrangling

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing categorical values
df['Dosage'] = df['Dosage'].fillna('Unknown')
df['Shipment Mode'] = df['Shipment Mode'].fillna(df['Shipment Mode'].mode()[0])

# Fill missing numerical values
df['Line Item Insurance (USD)'] = df['Line Item Insurance (USD)'].fillna(
    df['Line Item Insurance (USD)'].median()
)

# Convert date columns
df['Scheduled Delivery Date'] = pd.to_datetime(
    df['Scheduled Delivery Date'], errors='coerce'
)

df['Delivered to Client Date'] = pd.to_datetime(
    df['Delivered to Client Date'], errors='coerce'
)

# Check the cleaned dataset
df.info()
````

#### Chart - 8

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly as px

# Load Dataset
df = pd.read_csv('/content/SCMS_Delivery_History_Dataset.csv')

# Data Wrangling
df = df.drop_duplicates()
df['Dosage'] = df['Dosage'].fillna('Unknown')
df['Shipment Mode'] = df['Shipment Mode'].fillna(df['Shipment Mode'].mode()[0])
df['Line Item Insurance (USD)'] = df['Line Item Insurance (USD)'].fillna(
    df['Line Item Insurance (USD)'].median()
)
df['Scheduled Delivery Date'] = pd.to_datetime(
    df['Scheduled Delivery Date'], errors='coerce'
)
df['Delivered to Client Date'] = pd.to_datetime(
    df['Delivered to Client Date'], errors='coerce'
)

# Chart - 8 visualization code
# bivariate analysis
# shipment mode vs delivery delay
df['Delivery Delay'] = (
    df['Delivered to Client Date'] -
    df['Scheduled Delivery Date']
).dt.days
df.groupby('Shipment Mode')['Delivery Delay'].mean().sort_values().plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Delivery Delay by Shipment Mode')
plt.xlabel('Shipment Mode')
plt.ylabel('Average Delay (Days)')
plt.xticks(rotation=45)
plt.show()
````

##### 1. Why did you pick the specific chart?

It compares a categorical variable with a numerical variable.

##### 2. What is/are the insight(s) found from the chart?

Different shipment modes may have different average delivery delays.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

This is highly useful for transportation decisions. A shipment mode associated with higher delays may need investigation or reduced usage for time-sensitive deliveries.
